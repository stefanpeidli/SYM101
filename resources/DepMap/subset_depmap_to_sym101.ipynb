```python
import pandas as pd
import yaml
```

```python
all_data = pd.read_csv("/home/peidli/my_group_folder/data/DepMap/OmicsExpressionAllGenesTPMLogp1Profile.csv", index_col=0)

# From Profile ID to ModelID
ProfileID_to_ModelID = pd.read_csv("/home/peidli/my_group_folder/data/DepMap/OmicsProfiles.csv", index_col=0).ModelID.to_dict()
all_data.index = all_data.index.map(ProfileID_to_ModelID)

# From ModelID to Cell Line Name
with open('../../configuration/config.yaml', "r") as file:
    config = yaml.load(file, Loader=yaml.FullLoader)
lines_to_depmap_model_ids = config['lines_to_depmap_model_ids']
translation_dict = {x: y for y, x in lines_to_depmap_model_ids.items()}
all_data.index = all_data.index.map(translation_dict)

# Subset
all_data = all_data[all_data.index.notnull()]
gene_symbol = all_data.columns.str.split(" ").str[0]
gene_id = all_data.columns.str.split(" ").str[-1].str.replace(")", "", regex=False).str.replace("(", "", regex=False)
all_data.columns = pd.MultiIndex.from_tuples(list(zip(gene_symbol, gene_id)), names=["gene_symbol", "gene_id"])
all_data.to_csv("DepMap_SYM101_OmicsExpressionAllGenesTPMLogp1Profile.csv", index=True)
```
